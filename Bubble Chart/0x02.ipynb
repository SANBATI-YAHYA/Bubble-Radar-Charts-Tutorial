# Bubble chart virsion 1 

## Importer numpy et pyplot 

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Données fictives 

```python

city_names = ["Paris", "Lyon", "Marseille", "Nice", "Bordeaux", "Toulouse", "Nantes", 
"Strasbourg", "Lille", "Rennes"] 
sales = np.array([250, 180, 200, 150, 170, 190, 160, 140, 130, 120])  
population = np.array([2200, 1700, 1600, 900, 1200, 1400, 800, 600, 700, 500])
surface_area = np.array([105, 47, 240, 72, 49, 118, 65, 78, 35, 50]) * 10 
growth_index = np.random.rand(10)
colors = np.random.rand(10)
```

```python
plt.figure(figsize=(8,7))
plt.scatter(sales,population,sizes=surface_area,c=colors)
plt.colorbar()
```

## Créer un graphique à bulles 

```python

plt.figure(figsize=(8,7))
plt.scatter(sales,population,sizes=surface_area,c=colors) 
colorbar=plt.colorbar() #poure la bar de color 
colorbar.set_label('indexx_meryam') # pour le nom de bar color
plt.title('Graphique de vente et population')
plt.xlabel('sales')
plt.ylabel('population')
plt.grid()
for i, j,z  in zip(sales, population,city_names):
    
        plt.text(i, j, z, ha='right', color='green')
```
