# BUBBLE CHART

## Importer numpy et pyplot 

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Générer via randn de numpy  un tableau X aléatoire de 60 valeurs

```python
x=np.random.randn(60)
```

## Générer via randn de numpy  un tableau Y aléatoire de 60 valeurs

```python
y=np.random.randn(60)
```

## Générer via randn de numpy  un tableau de tailles de bulles aléatoire de 60 valeurs. Chaque 
### Valeur peut être dans l’intervalle [0,100] 

```python
sizes = np.abs(np.random.randn(60)) * 100 
```

## Générer via randn de numpy  un tableau de couleurs de 60 valeurs aléatoires

```python
colors = np.random.rand(60)
```

## Créer un graphique à bulles 

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=sizes, c=colors, alpha=0.7, cmap='viridis', label='Points de données') 
plt.show()
```

## Ajouter des étiquettes des axes, le titre 

```python
plt.title('Graphique de dispersion avec Annotations')
plt.xlabel('taille')
plt.ylabel('piods')
```

## Ajouter des annotations pour les points à la position [3,3], [4,4] 


```python

plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=sizes, c=colors, alpha=0.7, cmap='viridis', label='Points de données') 

plt.title('Graphique de dispersion avec Annotations')
plt.xlabel('taille')
plt.ylabel('piods')
plt.text(x[3],y[3],"yanos",ha='right',color='green')
plt.text(x[4],y[4],'yahya',ha='center',color='black')
plt.show()
```
